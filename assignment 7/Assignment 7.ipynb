#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.experience = experience

    def describe_employee(self):
        print(f"--- Employee Details ---"
              f"\nName: {self.name}"
              f"\nNumber: {self.number}"
              f"\nAddress: {self.address}"
              f"\nSalary: {self.salary}"
              f"\nPost: {self.post}"
              f"\nEducation: {self.education}"
              f"\nExperience: {self.experience} years"
              f"\nYear of Joining: {self.year_of_joining}\n")

    def update_year_of_experience(self, new_experience):
        self.experience = new_experience
        print(f"-> Updated experience for {self.name} to {self.experience} years.")

    def increase_salary(self, increament_amount):
        self.salary += increament_amount
        print(f"-> Increased salary for {self.name}. New Salary: {self.salary}")



employee101 = Employee(
    "Mughees", "03213213211", "Karachi", 200000, 
    "Software Developer", "2023", "Intermediate", 5
)

print("Name:", employee101.name)
print("Post:", employee101.post)
print("Salary:", employee101.salary)
print("Experience:", employee101.experience)
print()

employee101.describe_employee()
employee101.update_year_of_experience(6)
employee101.increase_salary(40000)
print("\n" + "="*40 + "\n")



employees = [
    Employee("Ali", "03001111111", "Lahore", 150000, "Data Analyst", "2022", "Bachelors", 3),
    Employee("Sara", "03002222222", "Islamabad", 250000, "Project Manager", "2020", "Masters", 7),
    Employee("Ahmed", "03003333333", "Karachi", 120000, "UI/UX Designer", "2023", "Bachelors", 2),
    Employee("Fatima", "03004444444", "Peshawar", 180000, "Backend Developer", "2021", "Bachelors", 4),
    Employee("Bilal", "03005555555", "Quetta", 95000, "QA Engineer", "2024", "Intermediate", 1),
    Employee("Zainab", "03006666666", "Faisalabad", 220000, "DevOps Engineer", "2019", "Masters", 6),
    Employee("Usman", "03007777777", "Multan", 130000, "Content Writer", "2023", "Bachelors", 3),
    Employee("Ayesha", "03008888888", "Rawalpindi", 300000, "Tech Lead", "2018", "Masters", 9),
    Employee("Hamza", "03009999999", "Hyderabad", 110000, "Support Engineer", "2024", "Bachelors", 1),
    Employee("Hira", "03001010101", "Sialkot", 175000, "Frontend Developer", "2022", "Bachelors", 4)
]

for emp in employees:
    emp.describe_employee()

employees[0].increase_salary(25000)
employees[1].update_year_of_experience(8)
employees[3].increase_salary(35000)
```
