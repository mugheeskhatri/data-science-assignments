```python
# Online Shopping Cart Program

# Nested dictionary storing product details
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6},
}

# Dictionary to store items in the cart: {product_id: quantity}
cart = {}


def resolve_product_id(raw_input):
    """Normalize product input such as 'P2', 'p2', or '2' into 'P2'."""
    cleaned = raw_input.strip().upper()
    if cleaned in products:
        return cleaned
    if cleaned.isdigit():
        candidate = f"P{cleaned}"
        if candidate in products:
            return candidate
    return None


def view_products():
    """Display all products in a clean, readable format."""
    print("\n--- Available Products ---")
    print(f"{'ID':<6} {'Product':<15} {'Price (Rs.)':<14} {'Stock':<6}")
    print("-" * 45)
    for pid, details in products.items():
        print(f"{pid:<6} {details['name']:<15} {details['price']:<14,} {details['stock']:<6}")
    print("-" * 45)


def add_to_cart():
    """Add a valid product and quantity to the cart while respecting stock limits."""
    view_products()

    # Loop until a valid product ID is entered or user types 'back'
    while True:
        raw_pid = input("\nEnter Product ID (e.g., P1 or 1, or 'back' to cancel): ").strip()
        if raw_pid.lower() == "back":
            print("Returning to main menu.")
            return

        pid = resolve_product_id(raw_pid)
        if pid is not None:
            break

        print("Invalid Product ID. Please enter a valid ID like P1 or 1.")

    available_stock = products[pid]["stock"]
    current_cart_qty = cart.get(pid, 0)
    remaining_stock = available_stock - current_cart_qty

    if remaining_stock <= 0:
        print(
            f"Sorry, no more units of {products[pid]['name']} are available to add "
            f"(you already have all {available_stock} in your cart)."
        )
        return

    # Loop until a valid quantity within remaining stock is entered or user types 'back'
    while True:
        raw_qty = input(
            f"Enter quantity (max {remaining_stock} available to add, or 'back' to cancel): "
        ).strip()

        if raw_qty.lower() == "back":
            print("Returning to main menu.")
            return

        try:
            qty = int(raw_qty)
        except ValueError:
            print("Invalid quantity. Please enter a whole number.")
            continue

        if qty <= 0:
            print("Quantity must be greater than 0.")
            continue

        if qty > remaining_stock:
            if current_cart_qty > 0:
                print(
                    f"Sorry, only {remaining_stock} unit(s) of {products[pid]['name']} "
                    f"are available to add ({current_cart_qty} already in your cart)."
                )
            else:
                print(
                    f"Sorry, only {remaining_stock} unit(s) of {products[pid]['name']} "
                    f"are available in stock."
                )
            continue

        cart[pid] = current_cart_qty + qty
        print(f"Added {qty} x {products[pid]['name']} to your cart. (Total in cart: {cart[pid]})")
        break


def view_cart():
    """Display cart items, subtotals, and the overall cart total."""
    if not cart:
        print("\nYour cart is currently empty.")
        return 0

    print("\n--- Your Shopping Cart ---")
    print(f"{'ID':<6} {'Product':<15} {'Price':<10} {'Qty':<6} {'Subtotal':<12}")
    print("-" * 55)

    cart_total = 0
    for pid, qty in cart.items():
        name = products[pid]["name"]
        price = products[pid]["price"]
        subtotal = price * qty
        cart_total += subtotal
        print(f"{pid:<6} {name:<15} Rs.{price:<7,} {qty:<6} Rs.{subtotal:<9,}")

    print("-" * 55)
    print(f"Cart Total: Rs. {cart_total:,}")
    return cart_total


def checkout():
    """Calculate final bill (with 10% discount for >= Rs. 5,000), update stock, and clear cart."""
    if not cart:
        print("\nYour cart is empty. Add items before checking out.")
        return

    cart_total = view_cart()

    if cart_total >= 5000:
        discount = cart_total * 0.10
        final_bill = cart_total - discount
        print(f"Discount (10%): -Rs. {discount:,.0f}")
    else:
        final_bill = cart_total
        print("Discount: Rs. 0 (Spend Rs. 5,000 or more for a 10% discount)")

    print(f"Final Bill: Rs. {final_bill:,.0f}")

    # Reduce product stock
    for pid, qty in cart.items():
        products[pid]["stock"] -= qty

    # Empty the cart for a new order
    cart.clear()
    print("\nCheckout complete! Stock has been updated and your cart is now empty.")


def main():
    """Main menu loop."""
    while True:
        print("\n=== Online Shopping Cart Menu ===")
        print("1. View Products")
        print("2. Add to Cart")
        print("3. View Cart")
        print("4. Checkout")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == "1":
            view_products()
        elif choice == "2":
            add_to_cart()
        elif choice == "3":
            view_cart()
        elif choice == "4":
            checkout()
        elif choice == "5":
            print("Thank you for shopping with us. Goodbye!")
            break
        else:
            print("Invalid choice. Please select an option from 1 to 5.")


# Run the program
main()
```
