Q11.
 You have a product code that follows the format "SKU-12345-XYZ".

Find the first position of the dash (-).

Count the number of dashes (-) in the product code.

Remove the "SKU-" prefix and "-XYZ" suffix from the product code, Extract the numeric part 12345

product_code = "SKU-12345-XYZ"

Hint: Use String Methods

```python
product_code = "SKU-12345-XYZ"
first_postition=product_code.find("-")
count_dash = product_code.count('-')
removed_prefix_suffix = product_code.replace("SKU-", "").replace("-XYZ", "")

print("First position", first_postition)
print("Dash count", count_dash)
print("Without prefix and suffix", removed_prefix_suffix)
```

Q14.
text = "    Data Analysis with Python is very powerful    "

Perform the following tasks:

Remove extra spaces using strip().
Convert the string into title case using title().
Replace "Powerful" with "Important".
Find the position of the word "Python".

```python
text_with_space = "    Data Analysis with Python is very powerful    "
stripped_text = text_with_space.strip()
titled_case = stripped_text.title()
replaced_text = titled_case.replace("Powerful", "Important")
python_position = stripped_text.find("Python")

print(stripped_text)
print(titled_case)
print(replaced_text)
print(python_position)
```
